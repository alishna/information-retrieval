```python
def tokenize_text(text):
    import nltk
    from nltk.tokenize import word_tokenize

    return word_tokenize(text)
```

```python
def remove_stopwords(tokens):
    import nltk
    from nltk.corpus import stopwords

    stop_words = set(stopwords.words('english'))

    filtered_tokens = [
        word for word in tokens
        if word.lower() not in stop_words
    ]

    return filtered_tokens
    print(filtered_tokens)
```

```python
from nltk.stem import PorterStemmer
def stemming(tokens):
    ps = PorterStemmer()
    return [ps.stem(word) for word in tokens]
```

```python

docs = {
    1: "It contains some text",
    2: "It also contains document"
}

# Inverted Index
inverted_index = {}

for doc_id, text in docs.items():
    Tokendata = tokenize_text(text)
    print("After Tokenization")
    print("doc", doc_id, ":", Tokendata)
    Sdata = remove_stopwords(Tokendata)
    print("After Stopword Removal")
    print("doc", doc_id, ":", Sdata)
    Sdata = stemming(Sdata)
    print("After Stemming")
    print("doc", doc_id, ":", Sdata)


    for word in Sdata:
        if word not in inverted_index:
            inverted_index[word] = {doc_id}
        else:
            inverted_index[word].add(doc_id)

print("Final Inverted Index:")
print(inverted_index)
```

```python
def search(query, index):
    tokens = tokenize_text(query)
    tokens = remove_stopwords(tokens)
    tokens = stemming(tokens)
    results = set()

    for word in tokens:
        if word in index:
            results.update(index[word])

    return results if results else "No documents found"


print(search("contain", inverted_index))
```
