```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('punkt')
nltk.download('stopwords')

docs = [
    "Information Retrieval is finding relevant documents",
    "Search engine uses algorithm to rank results",
    "Cosine similarity compares document vectors",
    "Machine Learning improves accuracy",
    "machine is useful for humans"
]
```

```python
stop_words = set(stopwords.words('english'))

def preprocess(doc):
    tokens = word_tokenize(doc.lower())
    return " ".join([w for w in tokens if w.isalnum() and w not in stop_words])

clean_docs = list(map(preprocess, docs))

print(clean_docs)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(clean_docs)
```

```python
query="machine learning"
```

```python
query_tokens = word_tokenize(query.lower())
query_filtered = [w for w in query_tokens if w.isalnum() and w not in stop_words]
query_clean = " ".join(query_filtered)
print(query_clean)
```

```python
query_vec = vectorizer.transform([query_clean])
```

```python
similarity = cosine_similarity(query_vec, tfidf_matrix)

for i, score in enumerate(similarity[0]):
    print(f"Doc {i+1}: {score:.2f}")

print("\nmatch:", docs[similarity.argmax()])
```
